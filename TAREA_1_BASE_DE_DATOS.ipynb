```python
con.sql("CREATE OR REPLACE TABLE enfermedades AS SELECT * FROM df")
con.sql("SELECT COUNT(*) AS total_filas FROM enfermedades").show()    #Cuéntame cuántas filas hay y le pone nombre a esa columna
```

```python
con.sql("SELECT * FROM enfermedades LIMIT 5").show()    #Muéstrame las primeras 5 filas de la tabla contratos, con todas sus columnas
```

```python
con.sql("SUMMARIZE enfermedades").show()  #max_rows=100
```

```python
resumen = con.sql("SUMMARIZE enfermedades").df()
resumen[["column_name", "min", "max", "null_percentage"]]
```

```python
# Nombres reales de las columnas
con.sql("DESCRIBE enfermedades").show()
```

```python
con.sql("SELECT COMUNA FROM contratos LIMIT 20").show()
```

```python
con.sql("SELECT SUM(Trasplantados) FROM contratos").show()
```

```python
# Porcentaje de nulos por columna, ordenado de mayor a menor
resumen = con.sql("SUMMARIZE enfermedades").df()
resumen[["column_name", "null_percentage"]].sort_values("null_percentage", ascending=False).head(10)
#Todo llenisimo
```

```python
# ¿El total de filas es igual al total de IDs unicos?
con.sql("""
    SELECT
        COUNT(*) AS total_filas,
        COUNT(DISTINCT COMUNA) AS comuna_unicos
    FROM enfermedades
""").show()

# Si total_filas > comuna_unicos, hay comunas repetidas
# en este caso, es logico, es una misma ciudad, mismo pais, etc. No lo tomaremos con error
```

```python
# Buscamos todas las formas distintas en que aparece "ROVIRA"
con.sql("""
    SELECT DISTINCT COMUNA
    FROM enfermedades
    WHERE COMUNA ILIKE '%ROVIRA%'
""").show() #El dataset parece estar limpio
```

```python
con.sql("""
    SELECT COMUNA, COUNT(*) AS veces
    FROM enfermedades
    GROUP BY COMUNA
    HAVING COUNT(*) > 1
    ORDER BY veces DESC
    LIMIT 10
""").show()

#La comuna mas comun en el dataset es la COMUNA ORIENTAL.
# Pero predomina la gente que no pertenece a ninguna comuna/barrio.
```
